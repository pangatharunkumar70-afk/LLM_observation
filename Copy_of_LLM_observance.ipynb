```python
import os
from google.colab import userdata

os.environ["LANGSMITH_TRACING"] = "true"
os.environ["LANGSMITH_API_KEY"] = userdata.get("LANGSMITH_API_KEY")
os.environ["LANGSMITH_PROJECT"] = "my-first-trace"
```

```python
from langchain.chat_models import init_chat_model
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

model = init_chat_model(
    "google_genai:gemini-3.6-flash",
    api_key=api_key
)

response = model.invoke("What is observability in software?")
print(response.content)
```

```python
import google.generativeai as genai
from google.colab import userdata

# Retrieve the same API key
api_key = userdata.get("GEMINI_API_KEY")

# Attempt to configure the genai library with the retrieved key
try:
    genai.configure(api_key=api_key)
    print("API key configured successfully (this does not guarantee it's valid for API calls).")
    # Attempt to list models to test authentication
    for m in genai.list_models():
        if "generateContent" in m.supported_generation_methods:
            print(m.name)
    print("\nIf you see models listed, the key might be okay but there's another issue.\nIf you get an error here, the key is definitely the problem.")
except Exception as e:
    print(f"Error configuring or using the API key directly with google.generativeai: {e}")
    print("Please ensure your 'GEMINI_API_KEY' secret contains a valid API key for the Gemini API.")
```
